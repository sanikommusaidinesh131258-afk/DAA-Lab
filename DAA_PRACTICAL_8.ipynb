IMPLEMENT OF GRAPH AND SEARCHING TREE

```python
# bfs
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'E'],
    'D': ['B', 'E'],
    'E': ['C', 'D']
}
from collections import deque

def bfs(graph, start):
    visited = {start}
    queue = deque([start])
    order = []

    while queue:
        node = queue.popleft()        # front of queue
        order.append(node)
        for nbr in graph[node]:
            if nbr not in visited:
                visited.add(nbr)      # mark on enqueue, not on dequeue
                queue.append(nbr)
    return order

print(bfs(graph, 'A'))    # ['A', 'B', 'C', 'D', 'E']
```

```python
# dfs

def dfs(graph, node, visited=None, order=None):
    if visited is None:
        visited, order = set(), []
    visited.add(node)
    order.append(node)
    for nbr in graph[node]:
        if nbr not in visited:
            dfs(graph, nbr, visited, order)
    return order

print(dfs(graph, 'A'))    # ['A', 'B', 'D', 'E', 'C']
```
